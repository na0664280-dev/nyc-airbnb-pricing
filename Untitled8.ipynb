## 2. أول نظرة
- الحجم: 48,895 صف و 16 عمود.
- الأنواع: price و minimum_nights و number_of_reviews و availability_365 أرقام صحيحة (int64). reviews_per_month رقم عشري. last_review نص (object) وهو في الأصل تاريخ.
- القيم الناقصة: last_review و reviews_per_month فيهم 10,052 صف ناقص لكل واحد (وهم نفس العدد، غالباً إعلانات بدون أي تقييم). name فيه 16 ناقص، و host_name فيه 21 ناقص.
- المكرر: 0 صف مكرر.
- الإحصائيات: سعر الليلة من 0 إلى 10,000 دولار، المتوسط حوالي 152 والوسيط 106. هذا يعني أن فيه قيم عالية جداً ترفع المتوسط. الحد الأقصى لـ minimum_nights هو 1,250 ليلة (أكثر من 3 سنوات، غير واقعي). availability_365 الوسيط 45 والربع الأول 0.
- القيم الفريدة: neighbourhood_group فيه 5 مناطق (Brooklyn, Manhattan, Queens, Staten Island, Bronx). room_type فيه 3 أنواع (Private room, Entire home/apt, Shared room).

```python
import matplotlib.pyplot as plt

# 1) توزيع السعر
df["price"].plot(kind="hist", bins=100)
plt.title("Price distribution")
plt.xlabel("price")
plt.show()

# 2) وسيط السعر حسب المنطقة
df[df.price < 500].groupby("neighbourhood_group")["price"].median().plot(kind="bar")
plt.title("Median price by borough")
plt.show()

# 3) وسيط السعر حسب نوع الغرفة
df[df.price < 500].groupby("room_type")["price"].median().plot(kind="bar")
plt.title("Median price by room type")
plt.show()

# أرقام نحتاجها للجمل تحت الرسمات
print("نسبة الأسعار تحت 500:", (df.price < 500).mean())
print(df[df.price < 500].groupby("neighbourhood_group")["price"].median())
print(df[df.price < 500].groupby("room_type")["price"].median())
```

أغلب الأسعار منخفضة: 97.4% من الإعلانات سعرها أقل من 500 دولار، وفيه عدد قليل جداً بأسعار عالية تصل إلى 10,000 دولار تسحب التوزيع لليمين وترفع المتوسط (152) فوق الوسيط (106).

Manhattan هي الأغلى بوسيط 143 دولار، وبعدها Brooklyn بـ 90، ثم Queens و Staten Island بـ 75 لكل منهما، وأرخص منطقة Bronx بوسيط 65 دولار. يعني سعر Manhattan أكثر من ضعف Bronx.

الشقة الكاملة (Entire home/apt) وسيطها 155 دولار، والغرفة الخاصة (Private room) 70، والغرفة المشتركة (Shared room) 45. يعني نوع الغرفة يغيّر السعر بشكل كبير، والشقة الكاملة أكثر من ضعف الغرفة الخاصة.

```python
print("price == 0:", (df.price == 0).sum())
print("price > 1000:", (df.price > 1000).sum())
print("min_nights > 365:", (df.minimum_nights > 365).sum())
print("availability_365 == 0:", (df.availability_365 == 0).sum())
print(df[["name","host_name","last_review","reviews_per_month"]].isna().sum())
```
