```python
import pandas as pd
df = pd.read_csv("AB_NYC_2019.csv")
print(df.shape)
```

```python
print("price = 0:", (df.price == 0).sum())
print("price > 1000:", (df.price > 1000).sum())
print("min_nights > 365:", (df.minimum_nights > 365).sum())
print("availability = 0:", (df.availability_365 == 0).sum())
print("no reviews:", df.last_review.isna().sum())
print("name null:", df.name.isna().sum(), "| host_name null:", df.host_name.isna().sum())
```

```python
import os
os.makedirs("data", exist_ok=True)

n0 = len(df)
df = df[df.price > 0]
df = df[df.price <= 1000]
df = df[df.minimum_nights <= 365]

df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
df["has_reviews"] = df["last_review"].notna().astype(int)
df = df.drop(columns=["last_review", "name", "host_name"])

print("قبل:", n0, "| حذفنا:", n0 - len(df), "| المتبقي:", len(df))
print("ناقص متبقي:", df.isna().sum().sum())
df.to_csv("data/clean.csv", index=False)
```

المشكلة	القرار	السبب بالرقم	الصفوف المتأثرة
price = 0	حذف	سعر صفر مستحيل، وprice هو الـ Target	11
price > 1000	حذف	239 صف فقط (أقل من 0.5%) وتشوّه الموديل	239
minimum_nights > 365	حذف	حجز أكثر من سنة غير منطقي	14
availability_365 = 0	نحتفظ، ونضيف is_unavailable في القسم 6	17,533 صف (36%)، حذفها يضيّع ثلث الداتا	0
reviews_per_month ناقص	تعبئة بصفر	10,052 ناقص، كلها بدون تقييمات	0 حذف
last_review ناقص	تحويل إلى has_reviews (0/1) ثم حذف العمود	10,052 ناقص	0 حذف
name / host_name ناقص	حذف العمودين	16 و21 فقط، وغير مفيدين للتسعير	0 صف

```python
df["is_unavailable"] = (df.availability_365 == 0).astype(int)
df["is_multi_host"] = (df.calculated_host_listings_count > 1).astype(int)
df["is_entire_home"] = (df.room_type == "Entire home/apt").astype(int)

for c in ["is_unavailable", "is_multi_host", "is_entire_home"]:
    print(c, df.groupby(c)["price"].median().to_dict())
```

```python
import matplotlib.pyplot as plt

cols = ["is_unavailable", "is_multi_host", "is_entire_home"]
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, c in zip(axes, cols):
    df.groupby(c)["price"].median().plot(kind="bar", ax=ax, color=["#4C72B0", "#DD8452"])
    ax.set_title(c)
    ax.set_xlabel("0 = لا ، 1 = نعم")
    ax.set_ylabel("median price ($)")
    ax.tick_params(axis="x", rotation=0)
plt.tight_layout()
plt.show()
```

**Feature Engineering: الأعمدة الجديدة**

1. `is_unavailable`: الإعلانات المتوقفة طول السنة وسيط سعرها 100$ مقابل 110$ للمتاحة، فالفرق 10$ فقط وتأثيره ضعيف.
2. `is_multi_host`: المضيفون أصحاب أكثر من إعلان وسيط سعرهم 91$ مقابل 117$ للأفراد، أي أرخص بنحو 22%.
3. `is_entire_home`: الشقة الكاملة وسيطها 160$ مقابل 70$ لغيرها، أي أكثر من الضعف، وهذا أقوى عمود من الثلاثة.
